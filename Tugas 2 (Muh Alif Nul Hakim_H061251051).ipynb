Prompt 1 — Ground Penetrating Radar (GPR) untuk Identifikasi Utilitas Bawah Tanah
R — Role
Anda adalah geofisikawan yang ahli dalam Ground Penetrating Radar (GPR) dan investigasi bawah permukaan dangkal.

T — Task
Analisis radargram GPR untuk mengidentifikasi reflektor bawah permukaan yang berpotensi menunjukkan keberadaan pipa, kabel, rongga, atau batas lapisan tanah.

C — Context
Survei dilakukan pada area kampus yang akan digunakan untuk pembangunan fasilitas baru. Informasi mengenai utilitas bawah tanah diperlukan agar kegiatan konstruksi tidak merusak instalasi yang sudah ada.

I — Input
Gunakan radargram, frekuensi antena, interval trace, panjang lintasan, posisi lintasan, dan informasi utilitas yang sudah diketahui jika tersedia.

K — Constraints

Jangan menyatakan suatu reflektor sebagai pipa atau kabel secara pasti tanpa bukti pendukung.

Bedakan refleksi berbentuk hiperbola dari reflektor lapisan yang menerus.

Pertimbangkan pengaruh frekuensi antena terhadap resolusi dan kedalaman penetrasi.

Jelaskan pengaruh tanah konduktif atau kondisi basah terhadap kualitas data.

O — Output
Hasil harus berisi: identifikasi pola reflektor, perkiraan posisi dan kedalaman target, penjelasan dasar interpretasi, kemungkinan jenis objek, faktor pengganggu, dan rekomendasi pemeriksaan lapangan.

Q — Quality Criteria
Interpretasi harus berbasis ciri radargram, menyebutkan ketidakpastian kedalaman, dan membedakan hasil observasi dari dugaan jenis objek.

V — Verification
Periksa kembali skala kedalaman, parameter kecepatan elektromagnetik yang digunakan, posisi hiperbola, serta cocokkan target dengan informasi utilitas yang tersedia.


````python
import numpy as np
import matplotlib.pyplot as plt

# 1. Membuat matriks latar belakang dengan noise (simulasi tanah)
# Ukuran: 100 sampel waktu/kedalaman x 200 titik trace/lintasan
gpr_data = np.random.normal(0, 0.1, (100, 200))

# 2. Membuat pola hiperbola (simulasi refleksi pipa bawah tanah)
# Posisi target diasumsikan berada di tengah lintasan (trace ke-100)
for i in range(200):
    # Rumus hiperbola untuk travel time GPR
    kedalaman = int(np.sqrt(15**2 + ((i - 100) * 0.4)**2))
    
    if kedalaman < 95: # Mencegah error index out of bounds
        # Menambahkan wavelet refleksi (puncak positif dan negatif)
        gpr_data[kedalaman:kedalaman+3, i] += 1.5  
        gpr_data[kedalaman+3:kedalaman+5, i] -= 1.0 

# 3. Visualisasi Radargram
plt.figure(figsize=(8, 5))
plt.imshow(gpr_data, aspect='auto', cmap='gray')

# 4. Pengaturan Tampilan Grafik
plt.title('Prompt 1: Radargram GPR\n(Identifikasi Pola Hiperbola Utilitas Bawah Tanah)', fontsize=12, fontweight='bold')
plt.xlabel('Posisi Lintasan (Nomor Trace)', fontsize=11)
plt.ylabel('Waktu Tempuh / Estimasi Kedalaman (ns)', fontsize=11)
plt.colorbar(label='Amplitudo Refleksi')

# Tampilkan grafik
plt.tight_layout()
plt.show()
````

Gambar yang dihasilkan adalah sebuah radargram yang menunjukkan pola lengkungan berbentuk hiperbola. Pola hiperbola ini sangat khas dalam interpretasi GPR karena gelombang elektromagnetik memancar dalam bentuk kerucut. Saat antena mendekati, melewati, dan menjauhi suatu objek berukuran kecil atau silindris (seperti pipa atau kabel utilitas), jarak tempuh gelombang akan membentuk kurva melengkung. Puncak dari hiperbola tersebut merepresentasikan posisi asli dan kedalaman dari objek target.

Anda adalah ahli geofisika hidrogeologi yang berpengalaman menggunakan metode Self Potential (SP).

Analisis distribusi potensial listrik alami pada permukaan untuk mengidentifikasi pola anomali yang mungkin berkaitan dengan pergerakan fluida bawah permukaan.

Pengukuran SP dilakukan pada daerah yang diduga memiliki rembesan air tanah menuju permukaan. Data digunakan untuk membantu memahami arah zona aliran fluida sebelum dilakukan investigasi hidrogeologi lanjutan.

I — Input
Masukkan koordinat titik pengukuran, beda potensial, elevasi, arah lintasan, serta informasi mata air atau rembesan yang telah diketahui.

K — Constraints

Jangan menganggap semua anomali SP berasal dari aliran air tanah.

Pertimbangkan efek elektroda, variasi topografi, mineralisasi, dan sumber potensial lainnya.

Jelaskan bahwa arah gradien potensial tidak selalu dapat langsung diterjemahkan sebagai arah aliran tanpa asumsi dan kontrol tambahan.

O — Output
Buat tabel data, peta atau deskripsi pola potensial, identifikasi zona gradien kuat, interpretasi kemungkinan jalur fluida, faktor alternatif penyebab anomali, serta kesimpulan.

````python
import numpy as np
import matplotlib.pyplot as plt

# 1. Membuat data lintasan (X) dari 0 hingga 100 meter
x_sp = np.linspace(0, 100, 50)

# 2. Membuat data anomali SP (Y) dalam satuan milivolt (mV)
# Menggunakan fungsi Gaussian terbalik untuk menyimulasikan anomali negatif 
# yang sering dikaitkan dengan aliran fluida bawah permukaan (rembesan)
# Ditambah dengan noise acak (np.random.normal) untuk menyimulasikan data lapangan yang realistis
sp_anomali = -40 * np.exp(-((x_sp - 50)**2) / 150) + np.random.normal(0, 1.5, 50)

# 3. Visualisasi Profil Self Potential
plt.figure(figsize=(8, 5))
plt.plot(x_sp, sp_anomali, marker='o', linestyle='-', color='blue', label='Data SP Terukur')

# 4. Menandai zona dugaan rembesan air tanah
plt.axvline(x=50, color='red', linestyle='--', linewidth=2, label='Zona Dugaan Rembesan Air')
plt.axvspan(40, 60, color='red', alpha=0.1, label='Area Gradien Kuat')

# 5. Pengaturan Tampilan Grafik
plt.title('Prompt 2: Profil Self Potential (SP)\n(Investigasi Aliran Air Tanah)', fontsize=12, fontweight='bold')
plt.xlabel('Jarak Lintasan (m)', fontsize=11)
plt.ylabel('Beda Potensial (mV)', fontsize=11)
plt.legend(loc='lower left')
plt.grid(True, linestyle=':', alpha=0.7)

# Tampilkan grafik
plt.tight_layout()
plt.show()
````

Grafik yang dihasilkan berupa kurva 1D yang menunjukkan penurunan drastis nilai beda potensial listrik membentuk anomali negatif (cekungan ke bawah). Dalam metode SP, aliran air tanah melalui pori-pori batuan menghasilkan potensial elektrokinetik (potensial aliran). Anomali negatif yang tajam pada grafik ini mengindikasikan lokasi zona aliran fluida yang cukup kuat, yang dalam konteks survei ini ditafsirkan sebagai titik rembesan air tanah ke permukaan.

Anda adalah geofisikawan eksplorasi mineral yang menguasai metode resistivitas dan Induced Polarization (IP).

T — Task
Interpretasikan kombinasi distribusi resistivitas dan chargeability untuk menentukan zona yang memiliki karakteristik geofisika yang dapat menjadi target eksplorasi mineralisasi.

C — Context
Survei dilakukan pada wilayah yang secara geologi memiliki indikasi mineralisasi sulfida. Data IP digunakan sebagai informasi tambahan untuk menentukan zona yang perlu diprioritaskan untuk pemeriksaan geologi atau pengeboran.

I — Input
Gunakan penampang resistivitas, penampang chargeability, koordinat lintasan, parameter konfigurasi elektroda, peta geologi, dan informasi singkapan mineral jika tersedia.

K — Constraints

Jangan menyamakan chargeability tinggi dengan endapan sulfida secara otomatis.

Pertimbangkan pengaruh lempung, porositas, fluida, dan jenis mineral lainnya.

Interpretasi harus menggunakan resistivitas dan chargeability secara bersama-sama.

Jangan menentukan kadar atau volume bijih dari data IP saja.

O — Output
Identifikasi zona anomali, klasifikasikan kombinasi resistivitas–chargeability, jelaskan kemungkinan penyebab geologinya, tentukan zona yang dapat menjadi target investigasi lanjutan, dan berikan tingkat keyakinan interpretasi.

Q — Quality Criteria

V — Verification
Periksa apakah zona chargeability tinggi berimpit dengan karakter resistivitas yang mendukung, apakah pola anomali menerus secara spasial, dan apakah sesuai dengan konteks geologi.

````python
import numpy as np
import matplotlib.pyplot as plt

# 1. Membuat grid spasial untuk jarak lintasan (X) dan kedalaman (Z)
# X dari 0 hingga 100 meter, Z dari 0 hingga 50 meter ke bawah
x_ip, z_ip = np.meshgrid(np.linspace(0, 100, 50), np.linspace(0, 50, 25))

# 2. Membuat data Chargeability sintetik (dalam satuan mV/V)
# Nilai latar belakang sekitar 5 mV/V. 
# Menambahkan anomali tinggi (hingga +30 mV/V) di koordinat tengah (x=50, z=25)
ip_data = 5 + 30 * np.exp(-((x_ip - 50)**2 + (z_ip - 25)**2) / 200)

# Menambahkan sedikit noise agar terlihat seperti hasil inversi data lapangan
ip_data += np.random.normal(0, 0.5, ip_data.shape)

# 3. Visualisasi Penampang 2D IP
plt.figure(figsize=(9, 5))
contour = plt.contourf(x_ip, z_ip, ip_data, cmap='jet', levels=20)

# 4. Pengaturan Tampilan Grafik
plt.colorbar(contour, label='Chargeability (mV/V)')
plt.gca().invert_yaxis() # Membalik sumbu Y karena ini adalah representasi kedalaman bawah permukaan

plt.title('Prompt 3: Penampang 2D Induced Polarization (IP)\n(Identifikasi Zona Anomali Mineralisasi Sulfida)', fontsize=12, fontweight='bold')
plt.xlabel('Jarak Lintasan (m)', fontsize=11)
plt.ylabel('Kedalaman (m)', fontsize=11)

# Menandai zona target eksplorasi
plt.plot(50, 25, marker='x', color='black', markersize=12, markeredgewidth=2, label='Target Pengeboran (Anomali Tinggi)')
plt.legend(loc='upper right')

# Tampilkan grafik
plt.tight_layout()
plt.show()
````

Gambar yang dihasilkan adalah peta kontur 2D (penampang bawah permukaan) dengan anomali chargeability tinggi di bagian tengah (ditandai dengan warna merah). Chargeability mengukur seberapa lama batuan dapat menyimpan muatan listrik setelah arus dimatikan. Zona berwarna merah ini mengindikasikan adanya material konduktif yang mampu mempolarisasi listrik, yang dalam konteks eksplorasi sering kali menjadi target utama keberadaan mineralisasi sulfida (bijih logam).

I — Input

K — Constraints

Periksa pengaruh topografi dan elevasi, konsistensi pengukuran berulang, besarnya anomali dibandingkan noise instrumen, serta apakah anomali tetap muncul setelah koreksi.

````python
import numpy as np
import matplotlib.pyplot as plt

# 1. Membuat data lintasan (X) dari 0 hingga 100 meter
x_grav = np.linspace(0, 100, 50)

# 2. Membuat data Anomali Gravitasi Residual (Y) dalam satuan miligal (mGal)
# Rongga memiliki densitas yang lebih rendah dari batuan sekitarnya (defisiensi massa),
# sehingga menghasilkan anomali gravitasi negatif.
# Kita gunakan fungsi Gaussian negatif ditambah noise instrumen/lingkungan.
grav_anomali = -0.08 * np.exp(-((x_grav - 50)**2) / 50) + np.random.normal(0, 0.003, 50)

# 3. Visualisasi Anomali Gravitasi Mikro
plt.figure(figsize=(8, 5))
plt.plot(x_grav, grav_anomali, marker='s', linestyle='-', color='green', label='Anomali Residual Terukur')

# 4. Menandai zona indikasi rongga bawah permukaan
plt.axvspan(42, 58, color='gray', alpha=0.2, label='Zona Indikasi Rongga (Densitas Rendah)')
plt.axvline(x=50, color='red', linestyle='--', linewidth=1.5, label='Pusat Anomali Minimum')

# 5. Pengaturan Tampilan Grafik
plt.title('Prompt 4: Anomali Gravitasi Mikro\n(Deteksi Rongga Bawah Permukaan)', fontsize=12, fontweight='bold')
plt.xlabel('Jarak Lintasan (m)', fontsize=11)
plt.ylabel('Anomali Residual (mGal)', fontsize=11)
plt.legend(loc='lower left')
plt.grid(True, linestyle=':', alpha=0.7)

# Tampilkan grafik
plt.tight_layout()
plt.show()
````

Grafik ini menampilkan kurva 1D dari anomali residual gravitasi dengan pola nilai yang merosot (anomali negatif) di tengah lintasan. Gravitasi mikro sangat sensitif terhadap perubahan massa jenis (densitas) di bawah permukaan dangkal. Penurunan nilai gravitasi ini mencerminkan adanya defisiensi massa (kehilangan massa), yang menjadi indikator kuat keberadaan struktur berdensitas rendah seperti rongga bawah tanah, terowongan, atau zona pelapukan yang intensif.

Prompt 5 — Seismik Refleksi untuk Pemetaan Struktur Bawah Permukaan
R — Role
Anda adalah ahli interpretasi seismik refleksi yang berpengalaman dalam pemetaan horizon dan struktur geologi bawah permukaan.

T — Task
Interpretasikan penampang seismik refleksi untuk memetakan horizon utama, sesar, lipatan, dan perubahan kontinuitas reflektor.

C — Context
Data seismik refleksi diperoleh untuk memahami struktur geologi pada kedalaman yang lebih besar. Penampang digunakan untuk membuat interpretasi awal struktur sebelum dikombinasikan dengan data sumur atau informasi geologi lainnya.

I — Input
Gunakan penampang seismik, skala waktu two-way travel time (TWT), informasi kecepatan jika tersedia, posisi lintasan, dan data sumur atau marker geologi apabila tersedia.

K — Constraints

Bedakan reflektor yang benar-benar terlihat dari interpretasi struktur.

Jangan mengubah TWT menjadi kedalaman tanpa model kecepatan yang jelas.

Jangan menentukan litologi hanya dari amplitudo refleksi.

Jelaskan kemungkinan multiple, noise, dan reflektor semu.

Jika data sumur tidak tersedia, nyatakan keterbatasan korelasi litologi.

O — Output
Berikan identifikasi horizon utama, interpretasi kontinuitas reflektor, kemungkinan lokasi sesar atau lipatan, penjelasan perubahan amplitudo, hubungan dengan geologi, tingkat keyakinan interpretasi, dan kesimpulan.

Q — Quality Criteria
Interpretasi harus konsisten secara struktur, mengikuti kontinuitas reflektor, membedakan observasi dari interpretasi, dan menyatakan ketidakpastian.

V — Verification
Periksa kembali kontinuitas horizon, kemenerusan sesar pada beberapa reflektor, konsistensi geometri struktur, serta kesesuaian interpretasi dengan data kecepatan atau sumur yang tersedia.

````python
import numpy as np
import matplotlib.pyplot as plt

# 1. Membuat matriks dasar untuk penampang seismik (Noise Latar Belakang)
# Ukuran: 100 sampel waktu (TWT) x 200 titik trace (CDP)
seismik_data = np.random.normal(0, 0.05, (100, 200))

# 2. Mensimulasikan Horizon (Lapisan Reflektor) dan Struktur Sesar
# Kita asumsikan ada sesar turun tepat di tengah (Trace ke-100)
for i in range(200):
    # Horizon 1 (Lapisan atas)
    kedalaman_h1 = 30 if i < 100 else 45
    # Horizon 2 (Lapisan bawah)
    kedalaman_h2 = 65 if i < 100 else 80
    
    # Menambahkan wavelet seismik (puncak dan palung) pada kedalaman horizon
    wavelet = np.sin(np.linspace(0, np.pi, 4))
    
    # Memasukkan wavelet ke dalam matriks data seismik
    seismik_data[kedalaman_h1:kedalaman_h1+4, i] += wavelet
    seismik_data[kedalaman_h2:kedalaman_h2+4, i] += wavelet * 1.5 # Reflektor lebih kuat

# 3. Visualisasi Penampang Seismik Refleksi
plt.figure(figsize=(10, 6))
# Menggunakan colormap 'seismic' yang umum digunakan di industri migas (merah-putih-biru)
plt.imshow(seismik_data, aspect='auto', cmap='seismic', vmin=-1.5, vmax=1.5)

# 4. Menandai Bidang Sesar (Interpretasi Struktur)
plt.axvline(x=100, color='black', linestyle='--', linewidth=2, label='Interpretasi Bidang Sesar')

# 5. Pengaturan Tampilan Grafik
plt.title('Prompt 5: Penampang Seismik Refleksi\n(Pemetaan Horizon dan Struktur Sesar)', fontsize=12, fontweight='bold')
plt.xlabel('CDP / Nomor Trace', fontsize=11)
plt.ylabel('Two-Way Time / TWT (ms)', fontsize=11)
plt.colorbar(label='Amplitudo Seismik')
plt.legend(loc='upper right')

# Tampilkan grafik
plt.tight_layout()
plt.show()
````

Visualisasi ini menunjukkan penampang seismik 2D dengan pola garis-garis mendatar yang merepresentasikan horizon atau batas lapisan batuan. Hal yang paling menonjol dari gambar ini adalah lapisan-lapisan tersebut terputus dan bergeser ke bawah tepat di tengah penampang. Keterputusan dan pergeseran reflektor (horizon) ini merupakan bukti atau interpretasi dari adanya sesar turun (patahan), di mana salah satu blok batuan telah bergerak relatif ke bawah akibat gaya tektonik.
