# Prompt 5 — Seismik Refleksi untuk Pemetaan Struktur Bawah Permukaan

```markdown
# R — Role
Anda adalah ahli interpretasi seismik refleksi yang berpengalaman dalam pemetaan horizon dan struktur geologi bawah permukaan.

# T — Task
Interpretasikan penampang seismik refleksi untuk memetakan horizon utama, sesar, lipatan, dan perubahan kontinuitas reflektor.

# C — Context
Data seismik refleksi diperoleh untuk memahami struktur geologi pada kedalaman yang lebih besar. Penampang digunakan untuk membuat interpretasi awal struktur sebelum dikombinasikan dengan data sumur atau informasi geologi lainnya.

# I — Input
Gunakan penampang seismik, skala waktu two-way travel time (TWT), informasi kecepatan jika tersedia, posisi lintasan, dan data sumur atau marker geologi apabila tersedia.

# K — Constraints
- Bedakan reflektor yang benar-benar terlihat dari interpretasi struktur.
- Jangan mengubah TWT menjadi kedalaman tanpa model kecepatan yang jelas.
- Jangan menentukan litologi hanya dari amplitudo refleksi.
- Jelaskan kemungkinan multiple, noise, dan reflektor semu.
- Jika data sumur tidak tersedia, nyatakan keterbatasan korelasi litologi.

# O — Output
Berikan identifikasi horizon utama, interpretasi kontinuitas reflektor, kemungkinan lokasi sesar atau lipatan, penjelasan perubahan amplitudo, hubungan dengan geologi, tingkat keyakinan interpretasi, dan kesimpulan.

# Q — Quality Criteria
Interpretasi harus konsisten secara struktur, mengikuti kontinuitas reflektor, membedakan observasi dari interpretasi, dan menyatakan ketidakpastian.

# V — Verification
Periksa kembali kontinuitas horizon, kemenerusan sesar pada beberapa reflektor, konsistensi geometri struktur, serta kesesuaian interpretasi dengan data kecepatan atau sumur yang tersedia.
```

````python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# 1. PARAMETER DAN INFORMASI AWAL
# ==========================================
frekuensi_antena = "500 MHz"
panjang_lintasan = 50.0  # dalam meter
interval_trace = 0.05    # dalam meter
kedalaman_maks = 5.0     # dalam meter

print("=======================================================")
print("METODE GPR - IDENTIFIKASI UTILITAS BAWAH TANAH SINTETIS")
print("=======================================================")
print(f"Frekuensi Antena        : {frekuensi_antena}")
print(f"Panjang Lintasan        : {panjang_lintasan} m")
print(f"Interval Trace          : {interval_trace} m")
print(f"Estimasi Kedalaman      : {kedalaman_maks} m")
print("Target                  : Pipa, kabel, rongga, batas lapisan")
print("=======================================================\n")

# ==========================================
# 2. PEMBUATAN DATA SINTETIS
# ==========================================
# Menentukan dimensi matriks data (Grid)
num_traces = int(panjang_lintasan / interval_trace) # Sumbu X (Jarak)
num_samples = 500 # Sumbu Y (Kedalaman/Waktu tempuh)

# Membuat background noise (Sinyal acak latar belakang)
data_gpr = np.random.normal(0, 0.1, (num_samples, num_traces))

# Mensimulasikan target Pipa/Kabel (Pola Hiperbola)
posisi_pipa_x = num_traces // 2  # Posisi pipa di tengah lintasan (25 m)
kedalaman_pipa_z = 150           # Indeks kedalaman (simulasi di ~1.5 m)

for i in range(num_traces):
    # Persamaan hiperbola sederhana
    jarak = (i - posisi_pipa_x) * interval_trace
    # Menghitung lengkungan hiperbola berdasarkan jarak
    z = int(np.sqrt(kedalaman_pipa_z**2 + (jarak * 50)**2)) 
    
    # Kondisi yang sudah diperbaiki agar tidak terjadi error broadcast
    if z >= 3 and z < (num_samples - 3):
        # Menambahkan anomali amplitudo tinggi untuk lengkungan hiperbola
        data_gpr[z-3:z+3, i] += np.sin(np.linspace(0, np.pi, 6)) * 2.5

# Mensimulasikan Batas Lapisan Tanah (Refleksi Datar)
batas_lapisan = 350 # Indeks kedalaman untuk batas lapisan (simulasi di ~3.5 m)
data_gpr[batas_lapisan-2:batas_lapisan+2, :] += 1.5

# ==========================================
# 3. VISUALISASI (RADARGRAM)
# ==========================================
plt.figure(figsize=(12, 6))

# Menggunakan colormap 'gray' (standar untuk radargram GPR)
plt.imshow(data_gpr, aspect='auto', cmap='gray', 
           extent=[0, panjang_lintasan, kedalaman_maks, 0])

plt.title(f'Simulasi Radargram GPR ({frekuensi_antena})', fontsize=14, fontweight='bold')
plt.xlabel('Jarak / Panjang Lintasan (m)', fontsize=12)
plt.ylabel('Kedalaman (m)', fontsize=12)
plt.colorbar(label='Amplitudo Pantulan')

# Anotasi untuk memperjelas target
plt.annotate('Target Pipa/Kabel (Hiperbola)', xy=(25, 1.5), xytext=(30, 1.0),
             arrowprops=dict(facecolor='red', shrink=0.05), color='red', weight='bold')
plt.annotate('Batas Lapisan Tanah', xy=(10, 3.5), xytext=(10, 4.2),
             arrowprops=dict(facecolor='blue', shrink=0.05), color='blue', weight='bold')

plt.tight_layout()
plt.show()
````
